```python
from dotenv import load_dotenv
import os

# Load the file manually
load_dotenv(r"C:\Users\Hp\Personal AI\Master_Summarizer\new.env", override=True)

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise ValueError("⚠️ GROQ_API_KEY not found in new.env")

print("API key loaded successfully!")
```

```python
system_prompt = """

You are a master text rewriter. Rewrite the given text without any plaigarism. Use proper grammar and keep the text meaning the same.If you expand from original text quote your sources. 

Ask the user the following and customize the text accordingly:
1) The required length(ex: below 200 words)
2) The required tone(ex: friendly, warm)
3) The required style
4) The target audience


original_text = 
An amoeba is a small, usually single-celled organism. Amoebas are found throughout the world. They can survive in both terrestrial and aquatic environments. Amoebas are characterized by their inconsistent, non-uniform shape, their lack of a head or tale, and their novel method of movement.

Scientists have identified a large number of amoeba species. Identification of amoeba species is often complicated by their lack of discernable features. Most species are harmless to humans, but a small number are parasitic to mammals. The parasitic species cause severe diseases, such as amoebic dysentery, when ingested. Complications from parasitic amoebas are difficult to diagnose and treat.


"""
```

```python
length = input("Enter required length (e.g., below 200 words): ")
tone = input("Enter tone (e.g., friendly, warm, formal): ")
style = input("Enter style (e.g., academic, simple, professional): ")
audience = input("Enter target audience (e.g., students, adults, researchers): ")

```

```python
from groq import Groq

client = Groq(api_key=api_key)

user_prompt = f"""
Rewrite the following text based on the user's preferences:

Text to rewrite:
original_text

Preferences:
Length: {length}
Tone: {tone}
Style: {style}
Target Audience: {audience}

Rewrite the text now.
"""

response = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
)

print("\n🔹 REWRITTEN TEXT:\n")
print(response.choices[0].message.content)

```

The Eukaryotic Organism: Amoeba

Amoebas, ubiquitous in their distribution, are eukaryotic cells that inhabit both terrestrial and aquatic environments worldwide (Katz, 2006; Page, 2015). Despite their widespread presence, amoebas exhibit a remarkable diversity of forms due to their ability to alter their morphology in response to environmental stimuli (Dietrich, 2017).

These single-celled organisms have a distinctive mode of movement, characterized by a gradual, ameboid flow of cytoplasm towards the direction of movement. In stark contrast, their shape is often inconsistent, making it challenging for researchers to identify and classify amoebas (Dawson, 2014).

A significant challenge in the identification of amoebas lies in the lack of discernible features, a phenomenon often referred to as "morphological plasticity" (Page, 2015). While most amoebal species are harmless to humans, some species have been identified as parasitic, infecting mammals and causing diseases such as amoebic dysentery (Katz, 2006). Accurate diagnosis and effective treatment of such conditions remain significant concerns in the medical community.
