```python
from dotenv import load_dotenv
import os

# Load the file manually
load_dotenv("new.env", override=True)

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise ValueError("⚠️ GROQ_API_KEY not found in new.env")

print("API key loaded successfully!")
```

```python
from groq import Groq

client = Groq(api_key=api_key)

prompt = """"

You are a very precise summarizer. Summarize the given text exactly into 2-3 sentances. Stick to the information given in the text only.Make sure each sentance is less than 50 words and captures relevant information.

text:
The Tesla Model 3 Long Range AWD delivers up to 629 km (WLTP) of driving range on a single charge, thanks to its dual-motor setup optimized for aerodynamics. The car accelerates from 0–100 km/h in 4.4 seconds, and it supports supercharging at up to 250 kW, allowing up to 282 km of range in only 15 minutes. Inside, the Model 3 features a wraparound interior design, premium materials, a 15.4" center touchscreen, an 8" rear touchscreen, and a 17-speaker audio system. The vehicle weighs around 1,824 kg, offers 682 litres of cargo space, and comes with a 4-year/80,000 km basic warranty plus an 8-year/192,000 km battery and drive unit warranty. 
Tesla
"""

response = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ]
)

print(response.choices[0].message.content)
```
